## 2. Sentencias  Condicionales `if`, `elif`, `else`
Control del flujo de ejecución del programa según las condiciones.


>**Nota:** Es útil pensar en una condición if como una habitación cerrada, si la condición es Verdadera podremos entrar y la aplicación podrá ejecutar algunas tareas predefinidas, pero si la condición es Falsa el programa ignorara dichas tareas.

```python
# Condicional simple if
age = 19
if age > 18:
    print("you can enter")

# Condicional if-else
age = 17
if age > 18:
    print("you can enter")
else:
    print("go see Meat Loaf")

# Condicional múltiple if-elif-else
age = 18
if age > 18:
    #parte con sangría, esta es la expresión que será ejecutada si la condición es verdadera
    print("you can enter")
elif age == 18:
    print("go see Pink Floyd")
else:
    print("go see Meat Loaf")
```

## 3. Operadores Lógicos:
`and`, `or`, `not` y Cortocircuito permiten combinar o modificar condiciones y las vuelven compuestas.

```python
album_year = 1983

# Operador and
if album_year > 1979 and album_year < 1990:
    print("El álbum fue lanzado en la década de los 80s")

# Operador or
if album_year < 1980 or album_year > 1989:
    print("El álbum NO fue lanzado en los 80s")
else:
    print("El álbum es de los 80s")

# Cortocircuito para evitar división por cero
denominador = 0
calculo_seguro = (denominador != 0) and (100 / denominador > 5)
print("Cálculo seguro con cortocircuito:", calculo_seguro)
```

> **Nota:** En una operación `and`, si el primer término es falso, Python no evalúa el segundo, evitando errores fatales en tiempo de ejecución.

## 5. Caso Práctico Aplicado: Motor de Reglas para Detección de Transacciones Sospechosas
Validación multicriterio de movimientos financieros.


```python
def evaluar_transaccion(monto: float, es_internacional: bool, intentos_fallidos: int) -> str:
    if intentos_fallidos >= 3:
        return "BLOQUEO: Múltiples intentos fallidos"
    elif monto > 10_000_000 or (es_internacional and monto > 2_000_000):
        return "ALERTA: Requiere autenticación de dos factores (2FA)"
    elif monto <= 0:
        return "RECHAZO: Monto inválido"
    else:
        return "APROBADA: Transacción regular"

print("Transacción normal:       ", evaluar_transaccion(150000, False, 0))
print("Transacción internacional:", evaluar_transaccion(3500000, True, 1))
print("Transacción bloqueada:    ", evaluar_transaccion(50000, False, 4))
```

> **Nota:** Los motores de reglas empresariales estructuran condicionales jerárquicos para filtrar y proteger datos transaccionales en tiempo real.

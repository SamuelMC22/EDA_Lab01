## 1. ¿Qué son los Diccionarios? y Definición
Un diccionario consiste en pares de llaves y valores encerrados entre llaves `{}` y separados por dos puntos `:`.


```python
# Diccionario con llaves de diferentes tipos (incluyendo tuplas inmutables)
Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), 'key5': 5, (0, 1): 6}

print("Diccionario Dict:", Dict)
print("Acceso a 'key1':", Dict["key1"])
print("Acceso con llave de tupla (0, 1):", Dict[(0, 1)])
```

## 2. Llaves y acceso a Datos
Consulta, adición de nuevas entradas y métodos `.keys()` y `.values()`.


```python
release_year_dict = {
    "Thriller": "1982", "Back in Black": "1980",
    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992",
    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976",
    "Saturday Night Fever": "1977", "Rumours": "1977"
}

# Consulta por llave
print("Año de lanzamiento de Thriller:", release_year_dict['Thriller'])

# Obtener todas las llaves y todos los valores
print("Llaves disponibles:", list(release_year_dict.keys()))
print("Valores disponibles:", list(release_year_dict.values()))

# Agregar una nueva entrada y eliminar otra
release_year_dict['Graduation'] = '2007'
del(release_year_dict['Thriller'])

# Verificar si una llave existe con in
print("¿'The Bodyguard' está en el diccionario?:", 'The Bodyguard' in release_year_dict)
print("¿'Thriller' sigue en el diccionario?:", 'Thriller' in release_year_dict)
```

> **Nota:**  A cada llave solo puede corresponderle un valor, sin embargo, múltiples llaves pueden tener el mismo. Las llaves solo pueden ser cadenas, números o tuplas, pero los valores pueden ser de cualquier tipo.


## 3. Acceso seguro con `.get()` y actualización con `.update()`
En el procesamiento de datos reales (archivos JSON, respuestas de APIs o bases NoSQL), es común encontrar registros incompletos o fuentes con estructuras heterogéneas.

#### Acceso Seguro con `.get(llave, valor_por_defecto)`
* El problema de los corchetes `dict[clave]`: Si la clave no existe, Python lanza inmediatamente un error `KeyError` y detiene el programa.
* La solución con `.get()`: Si la clave existe, devuelve su valor. Si no existe, devuelve `None` (o el valor de respaldo que definas) sin interrumpir el flujo.

#### Fusión y Actualización con `.update()`
* Permite actualizar valores de llaves ya existentes y agregar automáticamente las llaves nuevas que no estaban en el diccionario original en una sola operación.

Evitando excepciones de `KeyError`.

```python
# Acceso seguro con .get()
album_consultado = "Abbey Road"
anio = release_year_dict.get(album_consultado, "Año no registrado")
print(f"Consulta segura de '{album_consultado}': {anio}")

# Actualización en lote
release_year_dict.update({"Abbey Road": "1969", "Back in Black": "1980 (Remastered)"})
print(f"Año actualizado de Abbey Road: {release_year_dict.get('Abbey Road')}")
```

> **Nota:** `.get(clave, default)` evita que el programa se caiga si la clave no existe en el diccionario.

## 5. Caso Práctico Aplicado: Agregación de Frecuencias y Sumas (Simulación de `GROUP BY`)
Contador de transacciones por categoría.

```python
transacciones = [
    {"categoria": "Comida", "monto": 25000},
    {"categoria": "Transporte", "monto": 12000},
    {"categoria": "Comida", "monto": 45000},
    {"categoria": "Educacion", "monto": 150000},
    {"categoria": "Transporte", "monto": 8000}
]

resumen_gastos = {}
for t in transacciones:
    cat = t["categoria"]
    resumen_gastos[cat] = resumen_gastos.get(cat, 0) + t["monto"]

print("=== Resumen de Gastos por Categoría ===")
for cat, total in resumen_gastos.items():
    print(f"- {cat:<15}: ${total:,.2f} COP")
```

> **Nota:** El patrón `resumen[cat] = resumen.get(cat, 0) + monto` es la forma más eficiente y limpia en Python de calcular agregaciones por grupo.

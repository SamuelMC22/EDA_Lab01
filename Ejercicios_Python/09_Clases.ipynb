> **Nota:** En Python creamos una clase usando `class NombreClase(object):`. Cada método dentro de la clase recibe `self` como primer parámetro, que representa la instancia u objeto específico que se está utilizando.


## 2. Creación de una Clase y Métodos: La Clase `Circle`
Definición del constructor `__init__`, atributo de radio y color, método para sumar radio y método para graficar.


```python
class Circle(object):
    # Constructor con valores por defecto
    def __init__(self, radius=3, color='blue'):
        self.radius = radius
        self.color = color

    # Método para modificar el radio sumando un valor r
    def add_radius(self, r):
        self.radius = self.radius + r
        return self.radius

    # Método para dibujar el círculo
    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()

# 1. Crear el objeto RedCircle con radio 10 y color 'red'
RedCircle = Circle(10, 'red')

print("Radio de RedCircle:", RedCircle.radius)
print("Color de RedCircle:", RedCircle.color)

# 2. Modificar el radio directamente
RedCircle.radius = 1
print("Radio tras asignación directa:", RedCircle.radius)

# 3. Aplicar el método add_radius
RedCircle.add_radius(2)
print("Radio tras add_radius(2):", RedCircle.radius)

RedCircle.add_radius(5)
print("Radio tras add_radius(5):", RedCircle.radius)

# 4. Crear el objeto BlueCircle con radio 100
BlueCircle = Circle(radius=100)
print("Radio de BlueCircle:", BlueCircle.radius)
print("Color de BlueCircle:", BlueCircle.color)
```

> **Nota:** `RedCircle` y `BlueCircle` son objetos independientes. Modificar `RedCircle.radius` no cambia para nada el radio de `BlueCircle`.


## 3. La Clase `Rectangulo`
Modelado de rectángulos con atributos de ancho (`width`), alto (`height`) y color.


```python
class Rectangulo(object):
    # Constructor
    def __init__(self, width=2, height=3, color='r'):
        self.height = height
        self.width = width
        self.color = color

    # Método para dibujar el rectángulo
    def drawRectangle(self):
        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height, fc=self.color))
        plt.axis('scaled')
        plt.show()

# 1. Instancia SkinnyBlueRectangle (ancho 2, alto 10, color 'blue')
SkinnyBlueRectangle = Rectangulo(2, 10, 'blue')
print("SkinnyBlueRectangle -> Ancho:", SkinnyBlueRectangle.width, "| Alto:", SkinnyBlueRectangle.height, "| Color:", SkinnyBlueRectangle.color)

# 2. Instancia FatYellowRectangle (ancho 20, alto 5, color 'yellow')
FatYellowRectangle = Rectangulo(20, 5, 'yellow')
print("FatYellowRectangle  -> Ancho:", FatYellowRectangle.width, "| Alto:", FatYellowRectangle.height, "| Color:", FatYellowRectangle.color)
```

> **Nota:** La clase `Rectangulo` funciona como una plantilla donde podemos crear rectángulos altos y delgados o anchos pasando diferentes dimensiones al instanciarlos.


## 4. Quiz
**Enunciado:** La clase `matplotlib.patches.Ellipse` permite dibujar elipses desde matplotlib: `from matplotlib.patches import Ellipse`.
Se requiere crear una clase `Elipse` cuyo constructor acepte valores iniciales, y que contenga los siguientes 5 métodos:
1. Cambiar el ancho
2. Cambiar el alto
3. Cambiar el color de relleno
4. Cambiar el color de borde
5. Dibujar


```python
from matplotlib.patches import Ellipse

class Elipse(object):
    # Constructor con valores iniciales
    def __init__(self, width=6, height=3, facecolor='cyan', edgecolor='black'):
        self.width = width
        self.height = height
        self.facecolor = facecolor
        self.edgecolor = edgecolor

    # 1. Método para cambiar el ancho
    def cambiar_ancho(self, nuevo_ancho):
        self.width = nuevo_ancho
        return self.width

    # 2. Método para cambiar el alto
    def cambiar_alto(self, nuevo_alto):
        self.height = nuevo_alto
        return self.height

    # 3. Método para cambiar el color de relleno
    def cambiar_color_relleno(self, nuevo_color):
        self.facecolor = nuevo_color
        return self.facecolor

    # 4. Método para cambiar el color de borde
    def cambiar_color_borde(self, nuevo_color_borde):
        self.edgecolor = nuevo_color_borde
        return self.edgecolor

    # 5. Método para dibujar
    def dibujar(self):
        plt.gca().add_patch(Ellipse((0, 0), width=self.width, height=self.height,
                                    facecolor=self.facecolor, edgecolor=self.edgecolor, lw=2))
        plt.xlim(-self.width, self.width)
        plt.ylim(-self.height, self.height)
        plt.axis('scaled')
        plt.show()

# Creación de objeto de prueba
mi_elipse = Elipse(8, 4, 'purple', 'black')
print("Elipse inicial -> Ancho:", mi_elipse.width, "| Alto:", mi_elipse.height, "| Relleno:", mi_elipse.facecolor, "| Borde:", mi_elipse.edgecolor)

# Aplicamos los 4 métodos de cambio de atributos
mi_elipse.cambiar_ancho(12)
mi_elipse.cambiar_alto(6)
mi_elipse.cambiar_color_relleno('gold')
mi_elipse.cambiar_color_borde('navy')

print("Elipse modificada -> Ancho:", mi_elipse.width, "| Alto:", mi_elipse.height, "| Relleno:", mi_elipse.facecolor, "| Borde:", mi_elipse.edgecolor)
```

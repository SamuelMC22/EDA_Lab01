## 1. Definición y Tipos de Delimitadores
En Python, las cadenas son colecciones inmutables de caracteres delimitadas por comillas simples, dobles o triples.


```python
nombre_artista = "Michael Jackson"
registro_numerico = '1 2 3 4 5 6'
simbolos_especiales = '@#2_#]&*^%'

print(f"Cadena 1: {nombre_artista}")
print(f"Registro numérico como texto: '{registro_numerico}'")
print(f"Caracteres especiales: '{simbolos_especiales}'")
```

> **Nota:** Las cadenas almacenan texto puro, lo que hace que cualquier número o símbolo entre comillas adquiere tipo `str` y se contiene todos los métodos de procesamiento de texto.


## 2. Indexación Positiva y Negativa
Acceso  a caracteres específicos en la cadena de str.


```python
cadena = "Michael Jackson"

print(f"Índice 0 (primer carácter): '{cadena[0]}'")
print(f"Índice 5 (sexto carácter): '{cadena[5]}'")
print(f"Índice -1 (último carácter): '{cadena[-1]}'")
print(f"Índice -2 (penúltimo carácter): '{cadena[-2]}'")
```

## 3. Slicing y Stride
Permite obtener subcadenas usando la fórmula `[inicio:fin:paso]`.


>**Nota:** Slicing significa rebanado y stride hace referncia a paso

```python
cadena = "Michael Jackson"

primer_nombre = cadena[0:7]      # 'Michael'
apellido = cadena[8:15]     # 'Jackson'
paso_dos = cadena[::2]
cadena_invertida = cadena[::-1]     #Para invertir cadena

print(f"Primer nombre: '{primer_nombre}'")
print(f"Apellido: '{apellido}'")
print(f"Texto con paso 2: '{paso_dos}'")
print(f"Cadena invertida: '{cadena_invertida}'")
```

## 4. Concatenar y Replicar
Se usan los operadores  `+` y `*` para manipulación de secuencias.


```python
base = "Michael Jackson"
complemento = " is the best"

frase = base + complemento
divisor = "-" * 40    #Ayuda a adornar los textos y usa el operador *

print(divisor)
print(frase)
print(divisor)
print("Replicación de etiqueta:", 3 * "[DATA] ")
```

> **Nota:** Para concatenar varios elementos de texto se usa `''.join(lista)` que ya viene incluido en los métodos de los strings

Saltos de línea, tabulaciones y rutas de archivos.

## 6. Métodos Principales de Cadenas
`.upper()`, `.replace()`, `.find()`
Operaciones para transformación y búsqueda.


```python
cancion = "Thriller is the sixth studio album"

mayus = cancion.upper()
reemplazo = cancion.replace("Michael Jackson", "MJ")
indice_track = cancion.find("track")
indice_inexistente = cancion.find("Rock")

print(f"Original: {cancion}")
print(f"Mayúsculas: {mayus}")
print(f"Reemplazo: {reemplazo}")
print(f"Índice de 'track': {indice_track}")
print(f"Búsqueda fallida ('Rock'): {indice_inexistente} (retorna -1)")
```

> **Nota:** `.find()`  devuelve `-1` cuando no encuentra la subcadena, a diferencia de `.index()` que lanza una excepción `ValueError`.

## 8. Caso Práctico Aplicado: Parseo y Limpieza de Logs de Servidor
Extracción estructurada de campos desde registros de texto plano.


```python
log_raw = "   2026-08-13 10:15:30 [INFO] User:samuel.monsalve Status:200 Duration:45ms   "

log_clean = log_raw.strip()   #Quitamos los espacios
partes = log_clean.split(" ")    #Dividimos la cadena de texto para tenerla en partes

timestamp = f"{partes[0]} {partes[1]}"
nivel = partes[2].replace("[", "").replace("]", "")
usuario = partes[3].split(":")[1]  #tomo la parte 2 del split ([1])
status = int(partes[4].split(":")[1])
duracion_ms = int(partes[5].split(":")[1].replace("ms", ""))

print("=== Log  Estructurado ===")
print(f"Timestamp: {timestamp} | Nivel: {nivel}")
print(f"Usuario:   {usuario} | Status HTTP: {status} | Latencia: {duracion_ms} ms")
```

> **Nota:** En este ejemplo encontramos otros 3 operadores de cadenas `.strip()`, `.split()` y `.replace()` muy usados para exxtraer la información de los textos crudos

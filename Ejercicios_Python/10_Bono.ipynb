*Ejemplos:*
* `lesser_of_two_evens(2,4) --> 2`
* `lesser_of_two_evens(2,5) --> 5`

```python
def lesser_of_two_evens(a, b):
    # Verificamos si ambos son pares usando el operador módulo %
    if a % 2 == 0 and b % 2 == 0:
        return min(a, b)
    else:
        return max(a, b)

# Casos de prueba oficiales
print("Caso 1 (ambos pares 2,4):", lesser_of_two_evens(2, 4))
print("Caso 2 (un impar 2,5):   ", lesser_of_two_evens(2, 5))
```

### 2. GALLETAS DE ANIMALES
Escribir una función que toma una cadena de dos palabras y devuelve `True` si ambas palabras comienzan con la misma letra.

*Ejemplos:*
* `animal_crackers('Ciudad Calor') --> True`
* `animal_crackers('Lugar Frio') --> False`


```python
def animal_crackers(text):
    # Separamos la cadena en dos palabras usando .split()
    palabras = text.lower().split()
    # Comparamos la primera letra de la primera palabra con la primera letra de la segunda
    return palabras[0][0] == palabras[1][0]

# Comprobación de casos de prueba oficiales
print("animal_crackers('Ciudad Calor'):", animal_crackers('Ciudad Calor'))
print("animal_crackers('Lugar Frio'):  ", animal_crackers('Lugar Frio'))
```

> **Nota:** `.lower()` estandariza las mayúsculas/minúsculas antes de indexar `palabras[0][0] == palabras[1][0]`, evitando errores.


### 3. HACE VEINTE
Dados dos enteros, devuelve `True` si la suma de los enteros es 20 o si uno de los enteros es 20. De lo contrario, devuelve `False`.

*Ejemplos:*
* `makes_twenty(20,10) --> True`
* `makes_twenty(12,8) --> True`
* `makes_twenty(2,3) --> False`


```python
def makes_twenty(n1, n2):
    # Condición compuesta con operador lógico or
    return (n1 + n2 == 20) or (n1 == 20) or (n2 == 20)

# Comprobación de casos de prueba oficiales
print("makes_twenty(20, 10):", makes_twenty(20, 10))
print("makes_twenty(12, 8): ", makes_twenty(12, 8))
print("makes_twenty(2, 3):  ", makes_twenty(2, 3))
```

### 4. MAYÚSCULAS
Escribir una función que haga mayúscula la primera y cuarta letra de una cadena de texto.

*Ejemplos:*
* `old_macdonald('macdonald') --> MacDonald`
* *Nota:* `'macdonald'.capitalize()` retorna `'Macdonald'`.


```python
def old_macdonald(name):
    # Si la cadena es corta, capitalizamos normalmente
    if len(name) < 4:
        return name.capitalize()
    # Dividimos en dos segmentos [0:3] y [3:] y capitalizamos ambos con .capitalize()
    primera_mitad = name[:3].capitalize()
    segunda_mitad = name[3:].capitalize()
    return primera_mitad + segunda_mitad

# Comprobación de casos de prueba oficiales
print("old_macdonald('macdonald'):", old_macdonald('macdonald'))
```

### 5. REVERSA
Dada una oración, retornar la misma oración pero en reversa.

*Ejemplos:*
* `master_yoda('I am home') --> 'home am I'`
* `master_yoda('We are ready') --> 'ready are We'`
* *Nota:* Revisar método `.join()`: `" ".join(['Hello', 'world']) --> 'Hello world'`


```python
def master_yoda(text):
    # Separamos las palabras en una lista
    palabras = text.split()
    # Invertimos la lista de palabras usando slicing con paso negativo [::-1]
    palabras_invertidas = palabras[::-1]
    # Reensamblamos las palabras con un espacio entre ellas
    return " ".join(palabras_invertidas)

# Comprobación de casos de prueba oficiales
print("master_yoda('I am home'):   ", master_yoda('I am home'))
print("master_yoda('We are ready'): ", master_yoda('We are ready'))
```

> **Nota:** `palabras[::-1]` invierte el orden de las palabras de la lista (no las letras individuales), y `' '.join(...)` une la lista resultante en una nueva oración.

### 6. PROBLEMA 33
Dada una lista de enteros, retornar `True` si el arreglo tiene un 3 seguido por un 3.

*Ejemplos:*
* `has_33([1, 3, 3]) --> True`
* `has_33([1, 3, 1, 3]) --> False`
* `has_33([3, 1, 3]) --> False`


```python
def has_33(nums):
    # Recorremos hasta len(nums) - 1 para comparar parejas contiguas
    for i in range(0, len(nums) - 1):
        if nums[i] == 3 and nums[i+1] == 3:
            return True
    return False

# Comprobación de casos de prueba oficiales
print("has_33([1, 3, 3]):   ", has_33([1, 3, 3]))
print("has_33([1, 3, 1, 3]):", has_33([1, 3, 1, 3]))
print("has_33([3, 1, 3]):   ", has_33([3, 1, 3]))
```

> **Nota:** Iteramos hasta `len(nums) - 1` para no salirnos de los límites del arreglo al evaluar el índice siguiente `nums[i+1]`.


### 7. REPLICADOR
Dada una cadena, devuelve una cadena donde por cada carácter en el original hay tres caracteres REPETIDOS.

*Ejemplos:*
* `paper_doll('Hello') --> 'HHHeeellllllooo'`
* `paper_doll('Mississippi') --> 'MMMiiissssssiiissssssiiippppppiii'`


```python
def paper_doll(text):
    resultado = ""
    for char in text:
        resultado += char * 3
    return resultado

# Comprobación de casos de prueba oficiales
print("paper_doll('Hello'):      ", paper_doll('Hello'))
print("paper_doll('Mississippi'):", paper_doll('Mississippi'))
```

> **Nota:** Aprovechamos la replicación de cadenas de Python `char * 3` para triplicar cada carácter en un bucle.


### 8. BLACKJACK
**Enunciado:** Dados tres enteros entre 1 y 11, si su suma es menor o igual a 21, devuelve su suma. Si su suma excede 21 y hay un once, reduce la suma total en 10. Finalmente, si la suma (incluso después del ajuste) excede 21, devuelve 'BUST'.

*Ejemplos:*
* `blackjack(5,6,7) --> 18`
* `blackjack(9,9,9) --> 'BUST'`
* `blackjack(9,9,11) --> 19`


```python
def blackjack(a, b, c):
    total = sum([a, b, c])
    
    # Caso 1: Suma menor o igual a 21
    if total <= 21:
        return total
    # Caso 2: Suma excede 21 pero hay al menos un 11 (As)
    elif total > 21 and 11 in (a, b, c):
        total_ajustado = total - 10
        if total_ajustado > 21:
            return "BUST"
        return total_ajustado
    # Caso 3: Suma excede 21 sin Ases o tras ajuste
    else:
        return "BUST"

# Comprobación de casos de prueba oficiales
print("blackjack(5, 6, 7): ", blackjack(5, 6, 7))
print("blackjack(9, 9, 9): ", blackjack(9, 9, 9))
print("blackjack(9, 9, 11):", blackjack(9, 9, 11))
```

> **Nota:** En Blackjack, el As puede valer 11 o 1. Restar 10 equivale a convertir un As de valor 11 en un As de valor 1 para salvar la mano.

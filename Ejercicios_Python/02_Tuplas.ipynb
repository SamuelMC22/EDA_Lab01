#### *Objetivo del Cuaderno*
Comprender y aplicar el uso de tuplas en Python como estructuras de datos, usando la indexación, el slicing, el anidamiento y la ordenación segura de registros.


## 1. Definición y Creación de Tuplas
Las tuplas son estructuras de datos que almacenan una colección ordenada de elementos que se escriben con paréntesis `()` y están separados por comas.


```python
# Creación de tuplas
tuple1 = ("disco", 10, 1.2)
tupla_un_elemento = ("dato",)  # Requiere coma al final

print("Tupla 1:", tuple1)
print("Tipo de dato:", type(tuple1))
print("Tupla de un elemento:", tupla_un_elemento)
```

> **Nota:** La tupla de un solo valor debe llevar una coma `('dato',)` para que Python no la confunda con un paréntesis de operación matemática.


## 2. Indexación en Tuplas
Acceso a elementos mediante índices positivos y negativos.


```python
tupla = ("disco", 10, 1.2)

print(f"Índice 0: {tupla[0]}")
print(f"Índice 1: {tupla[1]}")
print(f"Índice 2: {tupla[2]}")
print(f"Índice -1 (último): {tupla[-1]}")
```

> **Nota:** La indexación funciona exactamente igual que en las cadenas, empezando desde `0` hasta `N-1`, o desde `-1` para el último elemento.


## 3. Concatenación y Slicing de Tuplas
Unión de tuplas con el operador `+` y extracción de sub-tuplas.


```python
tuple1 = ("disco", 10, 1.2)
tuple2 = tuple1 + ("hard rock", 10)

print("Tupla concatenada:", tuple2)

# Slicing de tuplas como en cadenas para crear subtuplas
sub_tupla = tuple2[0:3]
ultimos_dos = tuple2[3:5]

# Obtener la longitud de la tupla
print("Longitud de tupla 2:", len(tuple2))
print("Subtupla [0:3]:", sub_tupla)
print("Últimos dos [3:5]:", ultimos_dos)
```

> **Nota:** Al ser inmutables, concatenar o hacer slicing sobre una tupla siempre produce una nueva tupla, manteniendo la original intacta.


## 4. Inmutabilidad y Ordenamiento con `sorted()`
Demostración de inmutabilidad y generación de listas ordenadas.


```python
# Tupla numérica desordenada
C_tuple = (-5, 1, -3)

# Ordenamiento con sorted()
lista_ordenada = sorted(C_tuple)

print("Tupla original (permanece igual):", C_tuple)
print("Nueva lista ordenada:", lista_ordenada)
```

> **Nota:** `sorted()` recibe una tupla inmutable y retorna una **lista nueva** con los elementos ordenados, sin alterar la tupla base.


## 5. Tuplas Anidadas
Tuplas que contienen otras tuplas o listas en su interior.


```python
tupla_anidada = (1, 2, ("pop", "rock"), (3, 4), ("disco", (1, 2)))

print("Elemento 0:", tupla_anidada[0])
print("Elemento 2 (subtupla):", tupla_anidada[2])
print("Elemento anidado [2][1]:", tupla_anidada[2][1])
print("Acceso profundo [4][1][0]:", tupla_anidada[4][1][0])
```

> **Nota:** En tuplas anidadas usamos corchetes consecutivos `[i][j]` para entrar capa por capa en la estructura.


## 6. Quiz


```python
# Tupla ejemplo
genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock", "R&B", "progressive rock", "disco")

# 1. Encontrar la longitud de la tupla genres_tuple:
print("1. Longitud de genres_tuple:", len(genres_tuple))

# 2. Accede al elemento en el índice 3:
print("2. Elemento en índice 3:", genres_tuple[3])

# 3. Utiliza el slicing para obtener los índices 3, 4 y 5:
print("3. Slicing [3:6]:", genres_tuple[3:6])

# 4. Encuentra los primeros dos elementos de la tupla genres_tuple:
print("4. Primeros dos elementos [0:2]:", genres_tuple[0:2])

# 5. Encuentra el primer índice de 'disco':
print("5. Índice de 'disco':", genres_tuple.index("disco"))

# 6. Genera una lista ordenada desde la tupla C_tuple=(-5, 1, -3):
C_tuple = (-5, 1, -3)
print("6. Lista ordenada desde C_tuple:", sorted(C_tuple))
```

> **Nota:** `.index('disco')` busca la primera posición del valor. Si el elemento no estuviera en la tupla lanzaría error, a diferencia de cadenas que usa `.find()`.


## 7. Caso Práctico Aplicado: Tuplas como Claves Inmutables de Coordenadas Geográficas
Uso de tuplas para indexar matrices espaciales sin riesgo de modificación.


```python
# Coordenadas geográficas inmutables (Latitud, Longitud) como llaves de diccionario
puntos_interes = {
    (4.6097, -74.0817): "Plaza de Bolívar (Bogotá)",
    (4.6280, -74.0650): "Pontificia Universidad Javeriana",
    (6.2442, -75.5812): "Parque Explora (Medellín)"
}

ubicacion_consulta = (4.6280, -74.0650)
lugar = puntos_interes.get(ubicacion_consulta, "Ubicación no registrada")

print("=== Consulta de Puntos de Interés ===")
print(f"Coordenadas: {ubicacion_consulta} -> Sede: {lugar}")
```

> **Nota:** Las listas no se pueden usar como claves de diccionarios porque son mutables. Las tuplas sí son *hashables*, haciéndolas perfectas para coordenadas y pares de datos.


## Conclusiones del Cuaderno

1. Seguridad por inmutabilidad: Las tuplas protegen la integridad de registros fijos (como coordenadas o configuraciones) evitando cambios accidentales en el código.
2. Eficiencia de memoria: Al tener tamaño fijo y no permitir adiciones dinámicas, ocupan menos memoria en RAM y se procesan más rápido que las listas.
3. Capacidad Hashable: Su inmutabilidad permite utilizarlas como claves compuestas en diccionarios y conjuntos, algo imposible de hacer con listas.

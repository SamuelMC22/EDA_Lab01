## 1. Definición, Creación y Singleton Syntax
Las tuplas son secuencias inmutables que pueden almacenar elementos de cualquier tipo de dato.


```python
# Creación de tuplas
tupla_vacia = ()
tupla_singleton = ("Dato_Unico",) # Requiere coma final
registro = ("Sensor_01", 4.6097, -74.0817, 2640, True)

print(f"Tupla vacía: {tupla_vacia} (Longitud: {len(tupla_vacia)})")
print(f"Singleton: {tupla_singleton} (Tipo: {type(tupla_singleton).__name__})")
print(f"Registro heterogéneo: {registro}")
print(f"Tipo de dato: {type(registro)}")
```

> 💡 **Nota del autor:** La coma en `('Dato_Unico',)` es obligatoria; sin ella, Python evalúa los paréntesis como un agrupador de expresiones algebraicas y no como una tupla.


## 2. Indexación, Slicing y Tuplas Anidadas
Navegación posicional multidimensional y preservación de inmutabilidad.


```python
tupla_compleja = ("Samuel", (10, 20, 30), ["Bogota", "Cali"], 2026)

# Acceso a elementos anidados
print("Elemento 0:", tupla_compleja[0])
print("Subtupla anidada [1]:", tupla_compleja[1])
print("Elemento dentro de subtupla [1][2]:", tupla_compleja[1][2])
print("Elemento dentro de lista anidada [2][0]:", tupla_compleja[2][0])

# Modificación interna de objeto mutable contenido (permitido)
tupla_compleja[2].append("Medellin")
print("Tupla con lista interna actualizada:", tupla_compleja)
```

> 💡 **Nota del autor:** La tupla asegura que sus referencias internas permanezcan constantes. Si una referencia apunta a una lista mutable, el contenido de la lista puede mutar sin alterar la identidad del contenedor.


## 3. Desempaquetado Avanzado (*Extended Unpacking*)
Asignación declarativa de variables y captura de elementos residuales.


```python
registro_experimento = ("EXP-2026-A", "Temperatura", 23.4, 24.1, 23.8, 25.0, "Aprobado")

# Desempaquetado con operador *
id_exp, variable, *lecturas, estado = registro_experimento

print(f"ID Experimento: {id_exp}")
print(f"Variable:       {variable}")
print(f"Lecturas intermedias ({len(lecturas)} muestras): {lecturas}")
print(f"Estado Final:   {estado}")
```

> 💡 **Nota del autor:** El operador `*` captura un número arbitrario de elementos intermedios en una lista, simplificando el procesamiento de registros con longitud variable de métricas.


## 4. Métodos Nativos y Operaciones: `.count()` e `.index()`
Consultas de frecuencia y posición con garantía de no alteración.


```python
muestra_datos = (10, 25, 30, 25, 40, 25, 50, 60, 25)

# Conteo y búsqueda
total_25 = muestra_datos.count(25)
posicion_40 = muestra_datos.index(40)
pertenencia = 30 in muestra_datos

print(f"Muestra: {muestra_datos}")
print(f"Frecuencia del valor 25: {total_25} veces")
print(f"Primera posición de 40: Índice {posicion_40}")
print(f"¿Existe el valor 30 en la tupla?: {pertenencia}")
```

> 💡 **Nota del autor:** Al tener solo dos métodos (`count` e `index`), las tuplas presentan una sobrecarga de memoria significativamente inferior a las listas.


## 5. Quiz Oficial sobre Tuplas (Ejercicios del Profesor)
Resolución y verificación de operaciones sobre tuplas y ordenamiento.


```python
# Tupla de prueba del laboratorio
genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock", "R&B", "progressive rock", "disco")

# 1. Longitud de la tupla
longitud = len(genres_tuple)
print(f"1. Longitud de genres_tuple: {longitud}")

# 2. Acceso al elemento en índice 3
elem_3 = genres_tuple[3]
print(f"2. genres_tuple[3]: '{elem_3}'")

# 3. Slicing para obtener índices 3, 4 y 5
slice_3_5 = genres_tuple[3:6]
print(f"3. genres_tuple[3:6]: {slice_3_5}")

# 4. Primeros dos elementos
primeros_dos = genres_tuple[0:2]
print(f"4. genres_tuple[0:2]: {primeros_dos}")

# 5. Encontrar el primer índice de 'disco'
indice_disco = genres_tuple.index("disco")
print(f"5. genres_tuple.index('disco'): {indice_disco}")

# 6. Ordenar la tupla generando una nueva lista
tupla_numerica = (0, 5, 2, 8, 3, 1)
ordenada = sorted(tupla_numerica)
print(f"6. sorted((0, 5, 2, 8, 3, 1)): {ordenada}")
```

> 💡 **Nota del autor:** La función global `sorted()` acepta cualquier iterable inmutable (como tuplas) y retorna una nueva lista ordenada sin violar la inmutabilidad de la tupla original.


## 6. Caso Práctico: Tuplas como Claves Hashables en Grafos y Matrices
Indexación multidimensional segura en diccionarios.


```python
# Modelado de matriz de adyacencia de rutas logísticas
red_transporte = {
    ("Bogota", "Medellin"): {"distancia_km": 415, "tiempo_hrs": 8.5},
    ("Bogota", "Cali"):     {"distancia_km": 460, "tiempo_hrs": 9.2},
    ("Medellin", "Cali"):   {"distancia_km": 420, "tiempo_hrs": 8.0}
}

ruta_consulta = ("Bogota", "Cali")
info_ruta = red_transporte.get(ruta_consulta)

print(f"Consulta de Ruta {ruta_consulta[0]} -> {ruta_consulta[1]}:")
print(f"Distancia: {info_ruta['distancia_km']} km | Tiempo estimado: {info_ruta['tiempo_hrs']} hrs")
```

> 💡 **Nota del autor:** Las listas no pueden ser claves de diccionarios porque no implementan el método `__hash__`. Las tuplas inmutables sí son hashables y permiten crear claves compuestas naturales.


## 📌 Conclusiones del Cuaderno

1. **Seguridad por Inmutabilidad**: Las tuplas blindan la integridad de configuraciones fijas, constantes y registros transaccionales frente a escrituras no autorizadas.
2. **Eficiencia de Almacenamiento**: Python asigna memoria en bloques exactos para tuplas (*fixed allocation*), reduciendo el footprint en RAM comparado con listas dinámicas.
3. **Claves Compuestas Hashables**: Su capacidad hashable las convierte en la estructura por excelencia para indexar relaciones espaciales, temporales y grafos en memoria.

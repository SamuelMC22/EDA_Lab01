## 2. El bucle `for`
Estructura de control para ejecutar un bloque de código un número determinado de veces.


```python
dates = [1982, 1980, 1973]
N = len(dates)

# Bucle for iterando con índices
print("Iteración por índice")
for i in range(N):
    print(f"Índice {i}: {dates[i]}")

# Bucle for directamente sobre los elementos
print("Iteración directa sobre la lista")
for year in dates:
    print(f"Año: {year}")
```

> **Nota:** El ciclo hace directamente sobre la colección (`for year in dates:`) en Python; iterar por índice (`for i in range(N):`) se usa cuando necesitamos la posición para modificar la lista.


## 3. Modificación de Listas y `enumerate()` en Bucles `for`
Actualización de elementos e iteración simultánea con índice y valor.


```python
# Modificar elementos dentro de un bucle
squares = ['red', 'yellow', 'green', 'purple', 'blue']

for i in range(0, len(squares)):
    print("Before square", i, "is", squares[i])
    squares[i] = 'weight'
    print("After square", i, "is", squares[i])

# Iteración con enumerate (restaurando valores para probar)
squares = ['red', 'yellow', 'green', 'purple', 'blue']
print("--- Iteración elegante con enumerate() ---")
for i, square in enumerate(squares):
    print(f"Posición #{i} -> Elemento: {square}")
```

> **Nota:** `enumerate(lista)` devuelve tuplas `(indice, elemento)` en cada iteración, eliminando la necesidad de manejar contadores manuales como `i += 1`.


## 4. ¿Qué es el bucle `while`?
Ejecución iterativa controlada por una condición lógica que se evalúa antes de cada ciclo.

```python
dates = [1982, 1980, 1973, 2000]

i = 0
year = 0

while year != 1973:
    year = dates[i]
    i = i + 1
    print(f"Año examinado: {year}")

print("It took", i, "repetitions to get out of loop.")
```

> **Nota:** Es indispensable asegurar que dentro del bucle `while` la condición cambie hacia `False` (como incrementar `i = i + 1`), de lo contrario se produce un bucle infinito y se colpasan los recursos.

## 6. Caso Práctico Aplicado: Procesamiento por Lotes (*Batching*) de un Dataset
División eficiente de un dataset en paquetes de tamaño fijo.


```python
dataset_ids = list(range(101, 131)) # 30 registros
tamano_lote = 10

print(f"Total registros: {len(dataset_ids)} | Tamaño de lote: {tamano_lote}")
for i in range(0, len(dataset_ids), tamano_lote):
    lote = dataset_ids[i:i + tamano_lote]
    print(f"-> Procesando Lote [Índices {i}:{i+len(lote)}]: {lote}")
```

> **Nota:** El particionamiento en lotes con `range(0, total, batch_size)` es la técnica estándar para enviar datos a APIs externas o entrenar modelos en minilotes.

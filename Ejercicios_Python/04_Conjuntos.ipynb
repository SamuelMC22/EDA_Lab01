## 1. Contenido de un Conjunto y Creación
Los conjuntos son colecciones no ordenadas de elementos únicos que se pueden crear con llaves `{}` o con la función `set()` a partir de listas.

```python
# Creación directa con duplicados
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
print("Conjunto set1:", set1)

# Convertir una lista en un conjunto
album_list = ["Michael Jackson", "Thriller", 1982, "00:42:19", "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)
print("Conjunto desde album_list:", album_set)

# Convertir lista de géneros
music_genres = set(["pop", "pop", "rock", "folk rock", "hard rock", "soul", "progressive rock", "soft rock", "R&B", "disco"])
print("music_genres único:", music_genres)
```

> **Nota:** Al convertir una lista a `set()`, los duplicados se borran automáticamente. Los sets no mantienen un orden fijo de los elementos.


## 2. Operaciones con Conjuntos: `.add()`, `.remove()` y Membresía (`in`)
Estos comandos nos permiten modificar los conjuntos debido a su estrutura dinámica


```python
A = set(["Thriller", "Back in Black", "AC/DC"])
print("Conjunto inicial A:", A)

# Agregar elementos con .add()
A.add("NSYNC")
A.add("NSYNC") # El duplicado no genera error pero no se añade dos veces
print("A tras add('NSYNC'):", A)

# Eliminar con .remove()
A.remove("NSYNC")
print("A tras remove('NSYNC'):", A)

# Comprobar pertenencia con in
print("¿'AC/DC' está en A?:", "AC/DC" in A)
print("¿'Queen' está en A?:", "Queen" in A)
```

## 3. Operaciones Lógicas:
Intersección, Diferencia, Unión y Subconjuntos son operadores binarios y métodos nativos desde el álgebra de conjuntos.

```python
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])

print("Set 1:", album_set1)
print("Set 2:", album_set2)

# Intersección
interseccion = album_set1 & album_set2
print("Intersección:", interseccion)

# Diferencia
solo_set1 = album_set1.difference(album_set2)
solo_set2 = album_set2.difference(album_set1)
print("Solo en Set 1 (Set1 - Set2):", solo_set1)
print("Solo en Set 2 (Set2 - Set1):", solo_set2)

# Unión
union_sets = album_set1.union(album_set2)
print("Unión (|):", union_sets)

# Subconjuntos y Superconjuntos
print("¿album_set1 es superconjunto de album_set2?:", album_set1.issuperset(album_set2))
print("¿album_set2 es subconjunto de album_set1?:", album_set2.issubset(album_set1))
print("¿{'Back in Black', 'AC/DC'} es subconjunto de album_set1?:", {"Back in Black", "AC/DC"}.issubset(album_set1))
```

> **Nota:** Las operaciones lógicas de conjuntos equivalen a los cruces de tablas en bases de datos: `&` es un `INNER JOIN`, `|` es un `FULL OUTER JOIN` y `.difference()` es un `ANTI JOIN`.

```python
# 1. Transformar la lista ['rap','house','electronic music', 'rap'] en un conjunto:
lista_q1 = ['rap', 'house', 'electronic music', 'rap']
set_q1 = set(lista_q1)
print("1. Conjunto transformado:", set_q1)

# 2. Considera que la lista A = [1, 2, 2, 1] y el conjunto B = set([1, 2, 2, 1]). Compara sus sumas:
A_q2 = [1, 2, 2, 1]
B_q2 = set([1, 2, 2, 1])
print(f"2. sum(A) = {sum(A_q2)} (suma 1+2+2+1=6)")
print(f"2. sum(B) = {sum(B_q2)} (suma conjunto {B_q2} -> 1+2=3)")
print(f"2. ¿sum(A) == sum(B)?: {sum(A_q2) == sum(B_q2)}")

# 3. Crea un nuevo conjunto album_set3 que sea la unión de album_set1 y album_set2:
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])
album_set3 = album_set1.union(album_set2)
print("3. album_set3 (Unión):", album_set3)

# 4. Averigua si album_set1 es un subconjunto de album_set3:
es_sub = album_set1.issubset(album_set3)
print("4. ¿album_set1 es subconjunto de album_set3?:", es_sub)
```

> **Nota:** En la pregunta 2 vemos claramente la diferencia de los conjuntos: al eliminar duplicados, `sum(B)` solo suma `1 + 2 = 3`, mientras que la lista `sum(A)` suma `1 + 2 + 2 + 1 = 6`.


## 5. Caso Práctico Aplicado: Conciliación de IDs entre Bases de Datos
Detección de usuarios registrados vs usuarios con pagos activos.


```python
usuarios_crm = {"USR_01", "USR_02", "USR_03", "USR_04", "USR_05"}
usuarios_pagos = {"USR_02", "USR_04", "USR_06", "USR_07"}

activos_con_pago = usuarios_crm & usuarios_pagos
crm_sin_pago = usuarios_crm - usuarios_pagos
pagos_sin_registro_crm = usuarios_pagos - usuarios_crm

print("=== Conciliación de Usuarios ===")
print("Usuarios en CRM con pago al día:", activos_con_pago)
print("Usuarios en CRM sin pago (morosos):", crm_sin_pago)
print("Pagos de usuarios no registrados en CRM:", pagos_sin_registro_crm)
```

> **Nota:** Este caso muestra cómo los conjuntos resuelven tareas complejas de auditoría y cruce de datos con una sola línea de código.

## 1. ¿Qué es una Función? y Su beneficio

```python
# Función básica para sumar dos números
def add(a):
    """Suma 1 al número dado"""
    b = a + 1
    print(a, "if you add one is", b)
    return b

resultado = add(4)
print("Resultado retornado:", resultado)
```

> **Nota:** Al inicializar, la palabra `def` define la función, los nombres entre paréntesis son los parámetros formales, y `return` habla del valor que la función entrega a quien la llame.


## 2. Funciones Predefinidas de Python (`len`, `sum`, `sorted`)
Funciones integradas que vienen ya incluidas en pyton.


```python
album_ratings = [10.0, 8.5, 9.0, 7.0, 9.5]

print("Longitud len():", len(album_ratings))
print("Suma total sum():", sum(album_ratings))
print("Promedio calculado:", sum(album_ratings) / len(album_ratings))
print("Lista ordenada sorted():", sorted(album_ratings))
```

> **Nota:** En mejor usar las funciones que vienen predeterminadas que la reimplementación manual de algoritmos básicos.


## 3. Funciones con Condiciones y Bucles
Integración de lógica condicional e iterativa dentro de funciones.


```python
def type_of_album(artist, album, year_released):
    
    print(artist, album, year_released)
    if year_released > 1980:
        return "Modern"
    else:
        return "Oldie"
    
x = type_of_album("Michael Jackson", "Thriller", 1980)
print(x)
```

## 4. Argumentos por Defecto
Parámetros predeterminados cuando el usuario omite su valor.


```python
def isGoodRating(rating=4):
    if rating < 7:
        print("this album sucks it's rating is", rating)
    else:
        print("this album is good its rating is", rating)

# Invocación usando el valor por defecto
isGoodRating()

# Invocación con valor personalizado
isGoodRating(10)
```

> **Nota:** Los argumentos por defecto deben declararse siempre a la derecha de los argumentos obligatorios.


## 5. Variables globales y ámbito de una variable
Comportamiento de variables locales vs globales y uso de `global`.


```python
# Variable global

artist = "Michael Jackson"

def printer(artist):
    global internal_var 
    internal_var= "Whitney Houston"
    print(artist,"is an artist")

printer(artist) 
printer(internal_var)

# Modificación de variable global usando la palabra clave global
myFavouriteBand = "AC/DC"

def getBandRating(bandname):
    myFavouriteBand = "Deep Purple"
    if bandname == myFavouriteBand:
        return 10.0
    else:
        return 0.0

print("AC/DC's rating is:",getBandRating("AC/DC"))
print("Deep Purple's rating is: ",getBandRating("Deep Purple"))
print("My favourite band is:",myFavouriteBand)
```

> **Nota:** Una variable declarada dentro de una función tiene ámbito local y se destruye al terminar la ejecución de la función, para evitar repetir nombres.

## 7. Caso Práctico Aplicado: Módulo de Escalado Min-Max de Características
Función analítica reutilizable para normalizar matrices de datos.


```python
def normalizar_min_max(datos: list[float]) -> list[float]:
    """Escala un arreglo numérico al rango [0, 1]"""
    min_v = min(datos)
    max_v = max(datos)
    rango = max_v - min_v
    if rango == 0:
        return [0.0 for _ in datos]
    return [round((x - min_v) / rango, 4) for x in datos]

edades = [18.0, 21.0, 25.0, 30.0, 45.0, 60.0]
escaladas = normalizar_min_max(edades)

print("=== Normalización Min-Max ===")
print("Edades originales:", edades)
print("Edades normalizadas [0 a 1]:", escaladas)
```

> **Nota:** Encapsular algoritmos estadísticos en funciones modulares permite utilizarlas de manera consistente en todo el pipeline de Ciencia de Datos.
